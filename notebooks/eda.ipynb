```python
import pandas as pd

# Have a readable format and then load the file
pd.set_option("display.float_format", "{:,.2f}".format)
df = pd.read_csv("../data/processed/steam_games_analysis.csv")

# Initial data details
df.head()
df.info()
df.isna().sum()
df.describe()
```
